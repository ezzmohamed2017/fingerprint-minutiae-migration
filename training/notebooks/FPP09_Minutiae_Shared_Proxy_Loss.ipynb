# FPP09 — Minutiae Encoder: Shared Cross-Sensor Proxy Losses (incremental over FPP08; 3 repetitions)

## What this kernel does, in short
The minutiae → embedding model is the **FPP08 winner**: local structure, k = 24 neighbours, three neighbour layers, trained with partial-crop augmentation (8.73 % source-balanced validation EER; FingerJetFX + BOZORTH3: 6.51 %). Its weakest set is still **cross-sensor (DB1×DB2, 10.12 %)**. FPP09 changes **only the training loss** — minutiae only, **no images** — with the same two losses tested on the image model in FPP04L:

| Arm | Loss on every batch (64 pairs) | Plain meaning |
|---|---|---|
| `k24 … (FPP08)` (baseline) | cosine-BCE on pairs | the FPP08 winner, **re-scored from its saved outputs, not re-trained** |
| `bce_cosface` | cosine-BCE + **CosFace** on shared finger proxies (s 30, m 0.35) | every training finger has one learned **anchor** (proxy), the same for its DB1 and its DB2 templates; each template is pulled towards its own finger's anchor with a margin and pushed from all the others — whichever sensor captured it |
| `bce_proxyanchor` | cosine-BCE + **Proxy-Anchor** on the same shared proxies (α 32, δ 0.1) | the same anchors; each anchor pulls all its finger's templates in the batch and pushes the other fingers' templates |

**Why it may help:** the pair loss only sees two templates at a time; a shared anchor ties together **all** templates of a finger from **both** sensors, which is exactly the cross-sensor case. **After training the proxies are not part of the model** (the encoder is unchanged: same size, same speed); they are **saved** (`proxies.npy`) so a later kernel can continue exactly.

## How it is trained (incremental — nothing is re-run)
- Every run **starts from the FPP08 winner's weights of the same repetition** (same model, no layer changes; checked: the start model reproduces its saved embeddings and the new run starts with the same embeddings, cosine ≥ 0.999).
- **Proxies** start at the mean embedding of each training finger computed with the start model (as FPP04L), one proxy per finger (identity_id), shared by DB1 and DB2; λ = 1.0 (BCE + 1.0 × proxy loss), proxy norm parameterisation 0.01 (as FPP04L).
- **Same as FPP08:** data, the FPP08 winner's configuration and augmentation (incl. partial crop), batch 64 pairs, 12,000 pairs per epoch, source-balanced sampler; AdamW 3e-4, weight decay 1e-4, clip 1.0, ≤ 60 epochs; early stopping / LR drops / best epoch on the **source-balanced validation EER** (patience 10, LR × 0.3 after 3).

## What is reported
- **Decision — validation only:** winner = lowest mean source-balanced validation EER among the baseline and the two arms (ties within 0.01 points → higher AUC). Validation table (balanced + DB1 / DB2 / DB1×DB2: EER, AUC, F1, FMR, FNMR, mean ± std) with **FingerJetFX + BOZORTH3** and the **image model** as reference rows.
- **Test (MOLF balanced, DB1×DB2) and FingerDOS — reported for every row incl. F1, never used to decide.**
- Bias check with the 5 calibration subjects; rotation check per run; time reported only (the minutiae encoder runs once per enrolled template at migration).

**Hardware:** A100; 2 arms × 3 repetitions = 6 runs ≈ 2–3 h. **Needs:** FPP08 finished (winner manifest), FPP01–FPP07 outputs, fpp_lib ≥ 1.8.

## 2. Minutiae sets, protocol, shared finger classes, augmented pair sequence
Same templates and augmentation as FPP08. New: each training template gets the class of its **finger** (identity_id = subject | hand | finger), **the same class for its DB1 and DB2 templates** — these are the shared proxies' labels. Validation, test and FingerDOS never use labels.

```python
MINU = {d: M.load_minutiae(L.local_copy(ws['templates'] / f'{d}_fingerjetfx_templates.h5')) for d in L.DATASETS}
MOLF = ('MOLF_DB1', 'MOLF_DB2'); MINU_MOLF = MINU['MOLF_DB1'] + MINU['MOLF_DB2']; OFFSET = {'MOLF_DB1': 0, 'MOLF_DB2': len(MINU['MOLF_DB1'])}
SETSQ = {}
for q in (False, True):
    SETSQ[q] = {d: np.stack([M.to_set(m, quality=q) for m in MINU[d]]) for d in L.DATASETS}; SETSQ[q]['MOLF'] = np.concatenate([SETSQ[q][d] for d in MOLF])
for d in L.DATASETS:
    cnt = np.array([len(m['x']) for m in MINU[d]]); qq = np.concatenate([m['quality'] for m in MINU[d] if len(m['quality'])])
    print(f"{d}: {len(cnt)} templates | minutiae median {int(np.median(cnt))} | > 128: {100 * (cnt > 128).mean():.2f} % | quality mean {qq.mean():.3f}, range {qq.min():.2f}-{qq.max():.2f}")
IMG_MOLF = SETSQ[False]['MOLF']        # the validation callback embeds IMG_MOLF[bv.indices]; switched to the arm's input type in train_run
PROTO = {r: P.load_pairs(ws, r) for r in REPS}; print('FPP01 pair files verified for', len(REPS), 'repetitions')

def finger_classes(proto):
    # one class per TRAINING finger (identity_id), the same class for its DB1 and DB2 templates (FPP04L); -1 = not a training template
    lab = np.full(len(MINU_MOLF), -1, np.int64); keys = {}
    for d in MOLF:
        s = proto['samples'][d]; t_ = s[s.partition == 'train']
        for pos, key in zip(t_.sample_index.to_numpy(), t_.identity_id.astype(str).to_numpy()): lab[OFFSET[d] + int(pos)] = keys.setdefault(key, len(keys))
    sens = {k: set() for k in range(len(keys))}
    for d in MOLF:
        for k in np.unique(lab[OFFSET[d]:OFFSET[d] + len(MINU[d])]):
            if k >= 0: sens[int(k)].add(d)
    if sum(len(v) == 2 for v in sens.values()) != len(keys): raise RuntimeError('some training fingers lack one sensor: proxies would not be shared')
    return lab, len(keys)
CLASSES = {r: finger_classes(PROTO[r][0]) for r in REPS}
print({r[:6]: f'{k} shared finger proxies' for r, (_, k) in CLASSES.items()})

class SetPairSeq(tf.keras.utils.Sequence):
    # source-balanced pairs; each training template freshly augmented (+ partial crop if the model uses it), reproducible per (seed, epoch, batch); + the finger class of each template
    def __init__(self, sampler, seed, cfg, labels, workers=4):
        super().__init__(workers=workers, use_multiprocessing=False, max_queue_size=8); self.s = sampler; self.seed = seed; self.cfg = cfg; self.labels = labels; self.epoch = 0
    def __len__(self): return len(self.s)
    def _sets(self, idx, rng):
        if not TRAIN_AUGMENTATION: return SETSQ[self.cfg['quality']]['MOLF'][np.asarray(idx)]
        return np.stack([M.to_set(M.augment(MINU_MOLF[int(i)], rng, self.cfg['crop']), quality=self.cfg['quality']) for i in idx])
    def __getitem__(self, b):
        ia, ib, y = self.s.batch(b); rng = np.random.default_rng([self.seed, self.epoch, b])
        return (self._sets(ia, rng), self._sets(ib, rng), self.labels[np.asarray(ia)], self.labels[np.asarray(ib)]), y
    def on_epoch_end(self): self.s.on_epoch_end(); self.epoch += 1
```

## 3. Siamese model with shared proxies, balanced-EER callback, rotation check, start weights
The proxy loss is added to the pair loss during training only. `start_from(...)` loads the FPP08 winner (same repetition), checks it reproduces its saved embeddings and that the new run starts with the same embeddings.

```python
class EpochClock(tf.keras.callbacks.Callback):
    def on_epoch_begin(self, epoch, logs=None): self.t0 = time.perf_counter()
    def on_epoch_end(self, epoch, logs=None):
        if logs is not None: logs['epoch_seconds'] = time.perf_counter() - self.t0
class BalancedEERCallback(tf.keras.callbacks.Callback):
    # after each epoch: embed the validation images, score all validation pairs, source-balanced EER -> best checkpoint / LR drop / early stop (fpp_lib rule)
    def __init__(self, encoder, bv, ckpt, hist_csv):
        super().__init__(); self.encoder = encoder; self.bv = bv; self.ckpt = ckpt; self.hist_csv = hist_csv; self.x = IMG_MOLF[bv.indices]; self.rows = []
        self.stop = P.BalancedEERStopping(PATIENCE, LR_PATIENCE, LR_FACTOR)
    def _lr(self): return float(np.asarray(self.model.optimizer.learning_rate))
    def _set_lr(self, v):
        try: self.model.optimizer.learning_rate.assign(v)
        except AttributeError: self.model.optimizer.learning_rate = v
    def on_epoch_end(self, epoch, logs=None):
        logs = logs if logs is not None else {}; t0 = time.perf_counter()
        m = self.bv.score(L.l2n(T.embed(self.encoder, self.x, EMBED_BATCH))); logs.update(m); lr = self._lr(); d = self.stop.update(epoch, m['val_balanced_eer'], m['val_balanced_auc'])
        if d['improved']: self.model.save_weights(str(self.ckpt))
        if d['reduce_lr'] and lr > self.stop.min_lr: self._set_lr(max(lr * LR_FACTOR, self.stop.min_lr))
        if d['stop']: self.model.stop_training = True
        logs['validation_seconds'] = time.perf_counter() - t0
        self.rows.append({'epoch': epoch + 1, **{k: float(np.asarray(v)) for k, v in logs.items() if np.asarray(v).size == 1}, 'lr': lr, 'improved': d['improved']}); pd.DataFrame(self.rows).to_csv(self.hist_csv, index=False)
        print(f"  epoch {epoch + 1}: val balanced EER {100 * m['val_balanced_eer']:.2f} % | AUC {100 * m['val_balanced_auc']:.2f} % | "
              + ' '.join(f"{n.split('_')[0]} {100 * m[f'val_eer_{n}']:.2f}" for n in P.VAL_SOURCES) + (' *best' if d['improved'] else '') + (' | LR down' if d['reduce_lr'] else '') + (' | stop' if d['stop'] else ''))

class SetSiamese(tf.keras.Model):
    # shared encoder; loss = cosine-BCE (learned calibration) + LAMBDA_PROXY * proxy loss on the shared finger proxies (CosFace or Proxy-Anchor)
    def __init__(self, encoder, kind, n_classes, init_proxies, **kw):
        super().__init__(**kw); self.encoder = encoder; self.kind = kind; self.n_classes = n_classes
        self.calib = T.CosineCalibration(name='same_finger_probability'); self.bce = tf.keras.losses.BinaryCrossentropy()
        self.proxies = self.add_weight(name='finger_proxies', shape=(n_classes, M.EMBEDDING_DIM), dtype='float32', trainable=True, initializer='zeros')
        self.proxies.assign(np.asarray(init_proxies, np.float32) * PROXY_NORM)
        self.t_loss = tf.keras.metrics.Mean(name='loss'); self.t_bce = tf.keras.metrics.Mean(name='bce'); self.t_px = tf.keras.metrics.Mean(name='proxy_loss')
        self.t_acc = tf.keras.metrics.BinaryAccuracy(name='verification_accuracy'); self.t_auc = tf.keras.metrics.AUC(name='verification_auc')
    @property
    def metrics(self): return [self.t_loss, self.t_bce, self.t_px, self.t_acc, self.t_auc]
    def call(self, inputs, training=False):
        ea = self.encoder(inputs[0], training=training); eb = self.encoder(inputs[1], training=training)
        return self.calib(tf.reduce_sum(ea * eb, axis=1))
    def proxy_loss(self, E, lab):
        # E: L2-normalised embeddings of the 2 x 64 templates in the batch; lab: their finger classes (>= 0: every training template has one)
        W = tf.math.l2_normalize(self.proxies, axis=1); cos = tf.matmul(E, W, transpose_b=True); onehot = tf.one_hot(lab, self.n_classes, dtype=tf.float32)
        if self.kind == 'cosface':
            return tf.reduce_mean(tf.nn.softmax_cross_entropy_with_logits(onehot, COSFACE_S * (cos - COSFACE_M * onehot)))
        pos = tf.exp(-PA_ALPHA * (cos - PA_DELTA)) * onehot; neg = tf.exp(PA_ALPHA * (cos + PA_DELTA)) * (1.0 - onehot)       # Proxy-Anchor (Kim et al. 2020)
        with_pos = tf.cast(tf.reduce_sum(onehot, 0) > 0, tf.float32)
        lp = tf.reduce_sum(tf.math.log1p(tf.reduce_sum(pos, 0)) * with_pos) / tf.maximum(tf.reduce_sum(with_pos), 1.0)
        return lp + tf.reduce_sum(tf.math.log1p(tf.reduce_sum(neg, 0))) / float(self.n_classes)
    def train_step(self, data):
        (xa, xb, la, lb), y = data; y = tf.cast(tf.reshape(y, [-1]), tf.float32)
        with tf.GradientTape() as tape:
            ea = self.encoder(xa, training=True); eb = self.encoder(xb, training=True)
            p = self.calib(tf.reduce_sum(ea * eb, axis=1)); l_bce = self.bce(y, p)
            l_px = self.proxy_loss(tf.concat([ea, eb], 0), tf.cast(tf.concat([la, lb], 0), tf.int32)); loss = l_bce + LAMBDA_PROXY * l_px
        tv = self.trainable_variables; self.optimizer.apply_gradients([(g_, v_) for g_, v_ in zip(tape.gradient(loss, tv), tv) if g_ is not None])
        for m_, v_ in ((self.t_loss, loss), (self.t_bce, l_bce), (self.t_px, l_px)): m_.update_state(v_)
        self.t_acc.update_state(y, p); self.t_auc.update_state(y, p); return {m_.name: m_.result() for m_ in self.metrics}

def init_proxies(encoder, labels, n_classes, q):
    # FPP04L: the mean L2-normalised embedding of each shared finger class (DB1 + DB2 training templates, clean), L2-normalised
    idx = np.flatnonzero(labels >= 0); E = L.l2n(T.embed(encoder, SETSQ[q]['MOLF'][idx], EMBED_BATCH)); W = np.zeros((n_classes, M.EMBEDDING_DIM), np.float32)
    np.add.at(W, labels[idx], E); return L.l2n(W)

def build_set_siamese(encoder, lr, kind, n_classes, W0):
    m = SetSiamese(encoder, kind, n_classes, W0, name='set_siamese'); d = np.zeros((1,) + SETS_MOLF.shape[1:], np.float32); m((d, d)); m.built = True
    m.compile(optimizer=tf.keras.optimizers.AdamW(learning_rate=lr, weight_decay=WEIGHT_DECAY, clipnorm=CLIPNORM)); return m

ROT_IDS = [i for i in range(len(MINU['MOLF_DB1'])) if len(MINU['MOLF_DB1'][i]['x']) >= 10][:ROTATION_CHECK_TEMPLATES]
_rr = np.random.default_rng(1); ROT_PARAMS = [(_rr.uniform(-np.pi, np.pi), _rr.uniform(-20, 20), _rr.uniform(-20, 20)) for _ in ROT_IDS]
ROT = {q: (np.stack([M.to_set(MINU['MOLF_DB1'][i], quality=q) for i in ROT_IDS]), np.stack([M.to_set(M.rotate_translate(MINU['MOLF_DB1'][i], *p_), quality=q) for i, p_ in zip(ROT_IDS, ROT_PARAMS)])) for q in (False, True)}
def rotation_check(encoder, q):
    # cosine(embedding of a template, of the same template rotated +-180 deg and shifted +-20 px): the local structure must stay ~1
    a, b = ROT[q]; e0 = L.l2n(T.embed(encoder, a, EMBED_BATCH)); e1 = L.l2n(T.embed(encoder, b, EMBED_BATCH)); c_ = np.sum(e0 * e1, 1)
    return {'rotation_cos_mean': float(np.mean(c_)), 'rotation_cos_min': float(np.min(c_))}

def start_from(encoder, src_arm, rep, cfg):
    sd = run_dir(src_arm, rep)
    if not L.run_complete(sd): raise FileNotFoundError(f'{sd} not finished')
    scfg = cfg_of(src_arm); src = M.build_ls_encoder(scfg); src.load_weights(str(sd / 'encoder.weights.h5'))
    x_src, x_dst = SETSQ[scfg['quality']]['MOLF_DB1'][:32], SETSQ[cfg['quality']]['MOLF_DB1'][:32]
    E0 = L.load_npz(sd / 'embeddings_float32.npz')['MOLF_DB1'][:32]; E1 = L.l2n(T.embed(src, x_src, 32)); cs = float(np.min(np.sum(L.l2n(E0) * E1, 1)))
    if cs < 0.999: raise RuntimeError(f'{src_arm} {rep}: reloaded start weights do not reproduce the saved embeddings (min cosine {cs:.5f})')
    info = M.transfer(src, encoder); E2 = L.l2n(T.embed(encoder, x_dst, 32)); same = float(np.min(np.sum(E1 * E2, 1)))
    if same < 0.999: raise RuntimeError(f'new model does not start equal to {src_arm} (min cosine {same:.5f})')
    del src; return {'start_from': L.rel(ws, sd), 'start_check_cosine': round(cs, 6), 'start_equal_cosine': round(same, 6), 'layers_copied': info['layers_copied'],
                     'layers_grown_zero_rows': ' | '.join(info['layers_grown_zero_rows']), 'layers_new': ' | '.join(info['layers_new'])}
```

## 4. One run = one (arm, repetition), started from the FPP08 winner

```python
def train_run(arm, rep, start_arm):
    rd = KDIR / 'runs' / rep / arm
    if L.run_complete(rd): print(arm, rep, 'complete -> skipped'); return
    lrd = P.local_run_dir(KERNEL + ('_smoke' if QUICK_SMOKE_TEST else ''), rep, arm)
    proto, Pp = PROTO[rep]; seed = proto['seed']; tf.keras.backend.clear_session(); gc.collect(); T.set_all_seeds(seed); t = L.timer()
    global IMG_MOLF, SETS_MOLF
    cfg = cfg_of(arm); q = cfg['quality']; IMG_MOLF = SETS_MOLF = SETSQ[q]['MOLF']; SETS = SETSQ[q]      # inputs of this arm (with / without quality)
    encoder = M.build_ls_encoder(cfg); start = start_from(encoder, start_arm, rep, cfg); params = T.count_params(encoder)
    labels, n_cls = CLASSES[rep]; W0 = init_proxies(encoder, labels, n_cls, q); siamese = build_set_siamese(encoder, LR, ARMS[arm], n_cls, W0)
    ssf = float(proto['manifest']['same_subject_impostor_fraction'])
    train_seq = SetPairSeq(P.MixedPairs(P.sources(Pp, P.TRAIN_SOURCES, OFFSET), PAIR_BATCH, True, seed, TRAIN_PAIRS_PER_EPOCH, ssf), seed, cfg, labels)
    bv = P.BalancedValidation(Pp, OFFSET, max_pairs=VAL_PAIRS)
    LOG.add(step='train_start', arm=arm, repetition=rep, learning_rate=LR, loss=f'bce + {ARMS[arm]}', proxies=int(n_cls), **start, train_composition=json.dumps(train_seq.s.composition),
            validation_pairs=json.dumps(bv.composition), **params)
    ckpt = lrd / 'best_siamese.weights.h5'; hist_csv = lrd / 'training_history.csv'; fit_marker = lrd / 'fit_complete.json'
    if ckpt.is_file() and fit_marker.is_file():
        train_s = json.loads(fit_marker.read_text())['training_seconds']; print(arm, rep, 'fit finished earlier in this session -> evaluating the local checkpoint')
    else:
        cbs = [EpochClock(), BalancedEERCallback(encoder, bv, ckpt, hist_csv)]
        siamese.fit(train_seq, epochs=EPOCHS, callbacks=cbs, verbose=2); train_s = t(); fit_marker.write_text(json.dumps({'training_seconds': train_s}))
    siamese.load_weights(ckpt)
    hist = pd.read_csv(hist_csv); b_i, _ = P.BalancedEERStopping.replay(hist, EPOCHS, patience=PATIENCE, lr_patience=LR_PATIENCE, lr_factor=LR_FACTOR); best = hist.iloc[b_i]
    emb = {d: L.l2n(T.embed(encoder, SETS[d], EMBED_BATCH)) for d in L.DATASETS}                          # clean templates, inference mode
    encoder.save_weights(lrd / 'encoder.weights.h5'); np.savez_compressed(lrd / 'embeddings_float32.npz', **{d: e.astype(np.float32) for d, e in emb.items()})
    rot = rotation_check(encoder, q)
    np.save(lrd / 'proxies.npy', siamese.proxies.numpy())                     # training-only state, saved so a later kernel can continue exactly
    gpu_batch_ms, gpu_single_ms = T.time_encoder(encoder, SETS['MOLF_DB1'][:64], TIMING_REPEATS)
    extra = {'architecture': 'local_structure', **cfg, 'loss': f'bce + {ARMS[arm]}', 'proxies': int(n_cls), 'learning_rate': LR, **start, **rot, 'early_stopping': P.SELECTION_RULE, 'best_epoch': int(best.epoch), 'epochs_run': int(len(hist)),
             'val_balanced_eer_at_best': float(best.val_balanced_eer), 'val_balanced_auc_at_best': float(best.val_balanced_auc), 'train_auc_at_best': float(best.get('verification_auc', np.nan)),
             'training_seconds': train_s, 'seconds_per_epoch': float(hist['epoch_seconds'].mean()) if 'epoch_seconds' in hist else float('nan'), 'pair_batch': PAIR_BATCH,
             'embed_ms_per_template_gpu_batch64': gpu_batch_ms, 'embed_ms_single_template_gpu': gpu_single_ms, **params}
    S = {n: (np.asarray(Pp[n]['label'], int), L.pair_scores(emb[da], emb[db], Pp[n]).astype(np.float64)) for n, (da, db, _) in P.PAIR_SETS.items()}
    rows, thr = P.evaluate(S, KERNEL, arm, rep, extra)
    (lrd / 'encoder_config.json').write_text(json.dumps({'input': 'FingerJetFX minutiae set', 'architecture': 'local_structure', 'arm': arm, 'builder': 'fpp09_minutiae.build_ls_encoder(setup)', 'setup': cfg, 'start': start,
                                                         'local_structure': dict(edge1=M.EDGE1, edge2=M.EDGE2, point=M.POINT, geometry_scale=M.GEO_SCALE),
                                                         'coord_half': M.COORD_HALF, 'embedding_dim': M.EMBEDDING_DIM, 'dropout': M.DROPOUT, 'loss': f'cosine-BCE + {LAMBDA_PROXY} x {ARMS[arm]} on shared finger proxies (proxies.npy: training only, not deployed)', 'augmentation': dict(M.AUG, crop=cfg['crop']) if TRAIN_AUGMENTATION else None,
                                                         'learning_rate': LR, 'module': 'fpp09_minutiae.py'}, indent=2))
    P.save_run(lrd, S, rows, thr, {'kernel': KERNEL, 'arm': arm, 'repetition': rep, 'seed': seed, **extra})
    n_files = P.publish_run(lrd, rd, skip=('fit_complete.json', 'best_siamese.weights.h5'))
    P.fig_set_panel(f'{arm} {rep[:6]} (frozen validation-EER threshold)', {st: [(*P.group_scores(S, st), thr)] for st in P.PAPER_SETS}, FIG / 'runs' / rep / f'{arm}_panel.png')
    P.fig_training_balanced(hist, f'{arm} {rep[:6]} — stopping on source-balanced validation EER', FIG / 'runs' / rep / f'{arm}_training.png')
    v = [r_ for r_ in rows if r_['set'] == P.SELECTION_SET][0]
    LOG.add(step='train_done', arm=arm, repetition=rep, status='ok', minutes=round(t() / 60, 1), best_epoch=int(best.epoch), files_to_drive=n_files,
            val_balanced_eer=round(100 * v['eer'], 2), val_balanced_auc=round(100 * v['roc_auc'], 2), val_balanced_f1=round(100 * v['f1_macro'], 2), rotation_cos_mean=round(rot['rotation_cos_mean'], 4))
    del siamese, encoder, emb; gc.collect()
```

## 5. Train both arms (start: FPP08 winner of the same repetition)

```python
for arm in ARMS_TO_RUN:
    for rep in REPS_TO_RUN:
        try: train_run(arm, rep, BASE)
        except (tf.errors.ResourceExhaustedError, tf.errors.InternalError) as exc:
            LOG.add(step='train_failed', arm=arm, repetition=rep, status='gpu_error', error=repr(exc)[:400]); print('GPU error', arm, rep, repr(exc)[:200]); tf.keras.backend.clear_session(); gc.collect()
        except Exception as exc:
            LOG.add(step='train_failed', arm=arm, repetition=rep, status='error', error=repr(exc)[:400]); print('FAILED', arm, rep, exc)
print(LOG.frame().tail(12).to_string())
```

## 6c. CPU time, 1 thread, fresh process (reported only — the encoder is the same for every arm; the proxy loss does not change it)

```python
# probe -> decision for 1:1 from a fingerprint IMAGE:  minutiae path = FingerJetFX extraction + set conversion + embedding + cosine;  BOZORTH3 path = FingerJetFX + ISO->xyt + BOZORTH3.
# FingerJetFX and BOZORTH3 times are read from FPP04C/cpu_timing (same runtime type); everything below is measured here, 1 thread, after loading and warm-up.
WORKER = r'''
import os, sys, json, time, numpy as np
mod_dir, wfile, h5file, out, n_t, warm, embfile, cfg = sys.argv[1], sys.argv[2], sys.argv[3], sys.argv[4], int(sys.argv[5]), int(sys.argv[6]), sys.argv[7], json.loads(sys.argv[8])
os.environ['CUDA_VISIBLE_DEVICES'] = ''; os.environ['TF_CPP_MIN_LOG_LEVEL'] = '2'
for v in ('OMP_NUM_THREADS', 'OPENBLAS_NUM_THREADS', 'MKL_NUM_THREADS'): os.environ[v] = '1'
import tensorflow as tf
tf.config.threading.set_intra_op_parallelism_threads(1); tf.config.threading.set_inter_op_parallelism_threads(1)
sys.path.insert(0, mod_dir); import fpp09_minutiae as M
enc = M.build_ls_encoder(cfg); enc.load_weights(wfile); f = tf.function(lambda z: enc(z, training=False))
mins = M.load_minutiae(h5file)[:n_t]
E0 = np.load(embfile)['MOLF_DB1'][:16]; E1 = f(tf.constant(np.stack([M.to_set(m, quality=cfg['quality']) for m in mins[:16]]))).numpy()
check = float(np.min(np.sum(E0 / np.linalg.norm(E0, axis=1, keepdims=True) * E1, 1)))
if check < 0.999: raise SystemExit(f'CPU copy does not reproduce the saved embeddings (min cosine {check:.5f})')
for i in range(warm): f(tf.constant(M.to_set(mins[i % len(mins)], quality=cfg['quality'])[None])).numpy()
conv, lat = [], []
for m in mins:
    t = time.perf_counter(); s = M.to_set(m, quality=cfg['quality'])[None]; conv.append(1000 * (time.perf_counter() - t))
    x = tf.constant(s); t = time.perf_counter(); f(x).numpy(); lat.append(1000 * (time.perf_counter() - t))
g = np.random.default_rng(0).normal(size=(10000, M.EMBEDDING_DIM)).astype(np.float32); g /= np.linalg.norm(g, axis=1, keepdims=True); q = g[0].copy()
t = time.perf_counter()
for i in range(20000): float(q @ g[i % 10000])
cmp_ms = 1000 * (time.perf_counter() - t) / 20000
cpu = next((l.split(':', 1)[1].strip() for l in open('/proc/cpuinfo') if l.startswith('model name')), '')
json.dump({'templates_timed': len(lat), 'convert_ms_mean': float(np.mean(conv)), 'embed_ms_mean': float(np.mean(lat)), 'embed_ms_median': float(np.median(lat)),
           'embed_ms_std': float(np.std(lat)), 'embed_ms_p95': float(np.percentile(lat, 95)), 'cosine_1to1_ms': cmp_ms, 'check_cosine_vs_saved_embeddings': check, 'cpu': cpu}, open(out, 'w'))
'''
Path('/content/fpp09_cpu_worker.py').write_text(WORKER)
FX = P.kernel_dir(ws, 'FPP04C') / 'cpu_timing' / 'FPP04C_fingerjetfx_bozorth3_timing.csv'
fx = pd.read_csv(FX) if FX.is_file() else pd.DataFrame()
fjfx_ms = float(fx['fingerjetfx_ms_mean'].mean()) if len(fx) else float('nan'); xyt_ms = float(fx['iso_to_xyt_ms_mean'].mean()) if len(fx) else float('nan')
bz_ms = float(fx['bozorth3_ms_per_pair'].mean()) if len(fx) else float('nan'); cpu_ref = str(fx['cpu'].iloc[0]) if len(fx) and 'cpu' in fx else ''
CPU_ROWS = []; h5 = str(L.local_copy(ws['templates'] / 'MOLF_DB1_fingerjetfx_templates.h5'))
for a in complete:
    w = run_dir(a, REPS_TO_RUN[0]) / 'encoder.weights.h5'; out = f"/content/fpp09_cpu_{a.split(' ')[0]}.json"
    r = subprocess.run([sys.executable, '/content/fpp09_cpu_worker.py', '/content', str(w), h5, out, str(CPU_TIMING_TEMPLATES), str(CPU_TIMING_WARMUP), str(w.with_name('embeddings_float32.npz')), json.dumps(cfg_of(a))], capture_output=True, text=True)
    if r.returncode != 0: print(r.stderr[-3000:]); raise RuntimeError(f'CPU timing failed for {a}')
    d = json.load(open(out)); CPU_ROWS.append({'method': f'FingerJetFX + {a} encoder + cosine', 'template_creation_ms': fjfx_ms + d['convert_ms_mean'] + d['embed_ms_mean'],
                                               'fingerjetfx_ms': fjfx_ms, 'set_conversion_ms': d['convert_ms_mean'], 'embedding_ms': d['embed_ms_mean'], 'embedding_ms_p95': d['embed_ms_p95'],
                                               'comparison_1to1_ms': d['cosine_1to1_ms'], 'probe_to_decision_1to1_ms': fjfx_ms + d['convert_ms_mean'] + d['embed_ms_mean'] + d['cosine_1to1_ms'],
                                               'check_cosine_vs_saved_embeddings': d['check_cosine_vs_saved_embeddings'], 'cpu': d['cpu']})
CPU_ROWS.append({'method': 'FingerJetFX + BOZORTH3', 'template_creation_ms': fjfx_ms + xyt_ms, 'fingerjetfx_ms': fjfx_ms, 'set_conversion_ms': xyt_ms, 'embedding_ms': 0.0, 'embedding_ms_p95': np.nan,
                 'comparison_1to1_ms': bz_ms, 'probe_to_decision_1to1_ms': fjfx_ms + xyt_ms + bz_ms, 'cpu': cpu_ref})
cpu_fair = pd.DataFrame(CPU_ROWS); cpu_fair['threads'] = 1; cpu_fair['same_cpu_model'] = (cpu_fair.cpu == cpu_ref) if cpu_ref else np.nan
for i_, r_ in cpu_fair.iterrows():
    if r_.method != 'FingerJetFX + BOZORTH3':
        n = (r_.template_creation_ms - (fjfx_ms + xyt_ms)) / max(bz_ms - r_.comparison_1to1_ms, 1e-9); cpu_fair.loc[i_, 'break_even_impressions_vs_BOZORTH3'] = n
print('Fair CPU time (1 thread; FingerJetFX / BOZORTH3 from FPP04C cpu_timing):'); display(cpu_fair.round(4))
if cpu_ref and not cpu_fair.same_cpu_model.all(): print('NOTE: CPU model differs from the FPP04C BOZORTH3 timing runtime: re-run the FPP04C timing cells in this runtime for a like-for-like table')
```
